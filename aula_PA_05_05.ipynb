```python
# Escrever um programa que o usuário digita um número de 1 a 10 e o programa impri a tabuada do número digitado.

numero = int(input("Digite um número de 1 a 10 para ver a tabuada: "))

if 1 <= numero <= 10:
    print(f"tabuada do {numero}:")
    for i in range(1, 11):
        resultado = numero * i
        print(f"{numero} x {i} = {resultado}")
else:
    print("por favor, digite um número entre 1 e 10.")
```

```python
# 1 - Escrever um programa que o usuário digita um número e o programa imprimi até o número digitado.

numero = int(input("digite um número inteiro: "))

if numero > 0:
    print(f"contagem até {numero}:")
    for i in range(1, numero + 1):
        print(i)
else:
    print("por favor, digite um número inteiro positivo.")
```

```python
# 2 - Escrever um programa que o usuário digita um número e o programa imprimi o número digitado até o zero (contagem regressiva).


numero = int(input("digite um número inteiro positivo para a contagem regressiva: "))

if numero >= 0:
    print(f"contagem regressiva a partir de {numero}:")
    for i in range(numero, -1, -1):
        print(i)
else:
    print("por favor, digite um número inteiro não negativo.")
```

```python
# 3 - Escrever um programa que o usuário digita um número e o programa imprimi os números pares.

numero = int(input("digite um número inteiro: "))

if numero > 0:
    print(f"números pares até {numero}:")
    for i in range(2, numero + 1, 2):
        print(i)
elif numero == 0:
    print("não há números pares positivos até 0.")
else:
    print("por favor, digite um número inteiro positivo.")
```

```python
# 4 - Escrever um programa que o usuário digita um número e o programa imprimi os números impares.

numero = int(input("digite um número inteiro: "))

if numero > 0:
    print(f"números ímpares até {numero}:")
    for i in range(1, numero + 1, 2):
        print(i)
elif numero == 0:
    print("não há números ímpares positivos até 0.")
else:
    print("por favor, digite um número inteiro positivo.")
```

```python
# 5 - Escrever um programa que o usuário digita um número e programa soma os números até o número digitado.

numero = int(input("digite um número inteiro positivo: "))

if numero > 0:
    soma = 0
    for i in range(1, numero + 1):
        soma += i
    print(f"a soma dos números de 1 até {numero} é: {soma}")
elif numero == 0:
    print("a soma dos números até 0 é 0.")
else:
    print("por favor, digite um número inteiro positivo.")
```

```python
# 6 - Escrever um programa que o usuário digita uma palavra e o programa conta as letras da palavra.

palavra = input("digite uma palavra: ")

numero_letras = len(palavra)

print(f"a palavra '{palavra}' tem {numero_letras} letras.")
```

```python
# 7 - Escrever um programa que o usuário digita uma palavra e o programa conta as vogais da palavra.

palavra = input("digite uma palavra: ")

vogais = "aeiouAEIOU"

contador_vogais = 0
for letra in palavra:
    if letra in vogais:
        contador_vogais += 1

print(f"a palavra '{palavra}' tem {contador_vogais} vogal(is).")
```
