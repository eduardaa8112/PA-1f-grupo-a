```python
import random

# fazer um jogo de adivinhação

# o usuario ira digitar um numero de 0 a 10 para adivinhar o numero secreto
# o usuario tera 03 tentativas para acertar o numero
# o programa devera imprimir uma mensagem se o palpite do usuario e maior ou menor que o numero secreto
# o programa deevera imprimir uma mensagem de quente, morno ou frio

# regras

# se a diferença do palpite para o numero secreto for 03 - frio 02 - morno 01 - quente

numero_sec = random.randint(0, 10)
max_ten = 3
tentativas = 0

print("bem-vindo ao jogo de adivinhação!")

while tentativas < max_ten:
    try:
        palpite = int(input("digite seu palpite (entre 0 e 10): "))
    except ValueError:
        print("entrada inválida. Por favor, digite um número inteiro.")
        continue

    if not (0 <= palpite <= 10):
        print("por favor, digite um número entre 0 e 10.")
        continue

    if palpite == numero_sec:
        print(f"acertou! o número secreto era {numero_sec}. parabéns! :D")
        break
    else:
        tentativas += 1
        tentativas_restantes = max_ten - tentativas

        if tentativas_restantes > 0:
            diferenca = abs(palpite - numero_sec)

            if palpite < numero_sec:
                print("seu palpite é menor que o número secreto.")
            else:
                print("seu palpite é maior que o número secreto.")

            if diferenca == 1:
                print("você está bem quente!")
            elif diferenca == 2:
                print("você está morno.")
            else:
                print("você está frio.")

            print(f"você tem mais {tentativas_restantes} tentativa(s).")

if tentativas == max_ten and palpite != numero_sec:
    print(f"suas tentativas acabaram. o número secreto era {numero_sec}. tente novamente mais tarde.")
```
