```python
# calculadora de 4 operações

while True:
    print("calculadora de 4 Operações")
    print("1 : somar")
    print("2 : subtrair")
    print("3 : multiplicar")
    print("4 : divisão")
    print("5 : sair")

    try:
        opcao = int(input("escolha uma opção: "))
    except ValueError:
        print("entrada inválida. por favor, digite um número de 1 a 5.")
        continue

    if opcao == 5:
        print("obrigado por usar a calculadora! :D")
        break

    if not 1 <= opcao <= 4:
        print("opção inválida. por favor, escolha uma opção de 1 a 5.")
        continue

    try:
        num1 = float(input("digite o primeiro número: "))
        num2 = float(input("digite o segundo número: "))
    except ValueError:
        print("entrada inválida. por favor, digite números válidos.")
        continue

    if opcao == 1:
        resultado = num1 + num2
        print(f"Resultado da soma: {resultado}")
    elif opcao == 2:
        resultado = num1 - num2
        print(f"Resultado da subtração: {resultado}")
    elif opcao == 3:
        resultado = num1 * num2
        print(f"Resultado da multiplicação: {resultado}")
    elif opcao == 4:
        if num2 == 0:
            print("Erro: Não é possível dividir por zero.")
        else:
            resultado = num1 / num2
            print(f"Resultado da divisão: {resultado}")
```

```python
# jogo inteligente

# menu
# opção 1 : par ou impar
# opção 2 : positivo
# opção 3 : numero maior
# opção 4 : sair

while True:
    print("Jogo Inteligente ")
    print("Par ou Ímpar")
    print("Positivo, Negativo ou Zero")
    print("Número Maior")
    print("Sair")

    try:
        opcao = int(input("escolha uma opção: "))
    except ValueError:
        print("entrada inválida. Por favor, digite um número de 1 a 4.")
        continue

    if opcao == 1:
        # verificando se o numero do usuario é par ou ímpar
        try:
            numero = int(input("Digite um número inteiro: "))
            if numero % 2 == 0:
                print(f"o seu número inserido {numero} é pár. :D")
            else:
                print(f"o seu número inserido {numero} é ímpar. :D")
        except ValueError:
            print("entrada inválida. por favor, digite um número inteiro.")

    elif opcao == 2:
        # identificar se um numero é positivo, negativo ou zero.
        try:
            numero = int(input("digite um numero: "))
            if numero > 0:
                print(f"{numero} é positivo")
            elif numero < 0:
                print(f"{numero} é negativo")
            else:
                print(f"{numero} é zero")
        except ValueError:
            print("entrada inválida. por favor, digite um número inteiro.")

    elif opcao == 3:
        # maior de dois numeros
        # solicite dois numeros e mostre qual é o maior (ou se são iguais)
        try:
            num1 = float(input("digite o primeiro número: "))
            num2 = float(input("digite o segundo número: "))

            if num1 > num2:
                print(f"o maior é: {num1}!")
            elif num2 > num1:
                print(f"o maior é: {num2}!")
            else:
                print("Os números são iguais.")
        except ValueError:
            print("Entrada inválida. Por favor, digite números válidos.")

    elif opcao == 4:
        print("obrigado por jogar!")
        break

    else:
        print("opção inválida. Por favor, escolha uma opção de 1 a 4.")
```
