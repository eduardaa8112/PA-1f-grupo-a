```python
# Escrever um programa que faz uma contagem regressiva de 10 ate 0
# imprimir uma mensagem de acabou no final.
try:
    contador = int(input("digite um numero de 0 a 50:"))

    while contador >= 0:
        print(contador)
        contador -= 1
    print("acabou!")
except ValueError:
    print("digite apenas numeros!")
except Exception as e:
    print(f"ocorreu um erro inesperado: {erro}")
```

```python
# escrever um programa que recebe as 04 notas do bimestre e imprimi a media

nota = 0
bimestre = 1
resultado = 0

# processamento
while bimestre <= 4:
  nota = float (input(f"digite sua nota do {bimestre} bimestre"))
  bimestre +=1
  resultado = resultado + nota

if (resultado /4) >= 7:
  print(f"sua média foi {resultado/4}, voce esta aprovado")
elif (resultado/4) >= 5:
  print(f"sua média foi {resultado/4}, voce esta de recuperação")
else:
  print(f"sua média foi {resultado/4}, voce esta retido")
```

```python
saldo = 1000.00 # Saldo inicial !

while True:
    print(" MENU ")
    print("1. ver saldo")
    print("2. depositar")
    print("3. sacar")
    print("4. sair")

    try:
        opcao = int(input("escolha uma opção: "))
    except:
        print("entrada inválida. por favor, digite um número.")
        continue

    if opcao == 1:
        print(f"seu saldo atual é: R${saldo}")
    elif opcao == 2:
        try:
            deposito = float(input("digite o valor a depositar: R$"))
            if deposito > 0:
                saldo += deposito
                print(f"depósito de R${deposito} realizado com sucesso.")
                print(f"seu novo saldo é: R${saldo}")
            else:
                print("o valor do depósito deve ser positivo.")
        except:
            print("valor inválido.")
    elif opcao == 3:
        try:
            saque = float(input("digite o valor a sacar: R$"))
            if saque > 0:
                if saldo >= saque:
                    saldo -= saque
                    print(f"saque de R${saque} realizado com sucesso.")
                    print(f"seu novo saldo é: R${saldo}")
                else:
                    print("saldo insuficiente!")
            else:
                print("o valor do saque deve ser positivo.")
        except ValueError:
            print("valor inválido. por favor, digite um número.")
    elif opcao == 4:
        print("obrigado por usar o Banco! :D")

    else:
        print("opção inválida. Por favor, escolha uma opção de 1 a 4.")
```
